# PCA vs. Factor Analysis
:label:`sec_factor`

Factor analysis encompasses a variety of techniques used for uncovering the underlying factors or latent variables that explain the observed data. Each method makes different assumptions about the data and employs different statistical models. This chapter covers Probabilistic Principal Component Analysis (PPCA), Gaussian Matrix Factorization, and Poisson Matrix Factorization.

### Probabilistic Principal Component Analysis (PPCA)

### Overview

- **PPCA** is a probabilistic approach to PCA that frames the dimensionality reduction problem in a probabilistic context.
- It models the data as being generated by a set of latent variables with added Gaussian noise.

### Model

The data $\mathbf{x}_i$ is modeled as:

$$
\mathbf{x}_i = \mathbf{W}\mathbf{z}_i + \boldsymbol{\mu} + \boldsymbol{\epsilon}_i
$$

where:

- $\mathbf{x}_i$ is the observed data.
- $\mathbf{W}$ is the factor loading matrix.
- $\mathbf{z}_i$ is the latent variable or factor.
- $\boldsymbol{\mu}$ is the mean vector.
- $\boldsymbol{\epsilon}_i$ is Gaussian noise with covariance $\sigma^2\mathbf{I}$.

### Characteristics

- **Latent Variables:** The latent variables are modeled as $\mathbf{z}_i \sim \mathcal{N}(0,\mathbf{I})$.
- **Maximum Likelihood:** Parameters $\mathbf{W}$ and $\sigma^2$ are estimated using maximum likelihood.
- **Principal Components:** The principal components in PCA correspond to the maximum-likelihood solution in PPCA.

### Gaussian Matrix Factorization

### Overview

Gaussian Matrix Factorization is used to decompose a matrix into lower-dimensional latent factors under the assumption of Gaussian noise. It is commonly used in collaborative filtering and recommendation systems.

### Model

The data matrix $\mathbf{X}$ is decomposed into two lower-dimensional matrices:

$$
\mathbf{X} \approx \mathbf{U}\mathbf{V}^T
$$

where:

- $\mathbf{U} \in \mathbb{R}^{n \times k}$ is the matrix of latent factors for rows.
- $\mathbf{V} \in \mathbb{R}^{m \times k}$ is the matrix of latent factors for columns.
- $k$ is the number of latent factors.

The entries of $\mathbf{X}$ are assumed to be Gaussian-distributed around $\mathbf{U}\mathbf{V}^T$.

### Characteristics

- **Gaussian Noise:** The model assumes Gaussian noise in the observed data.
- **Latent Factors:** It captures latent factors that explain the observed data.
- **Optimization:** It can be solved using alternating least squares or stochastic gradient descent.

### Poisson Matrix Factorization

### Overview

Poisson Matrix Factorization is used for decomposing count data into latent factors. It is commonly used for applications such as topic modeling and collaborative filtering with count data.

### Model

The data matrix $\mathbf{X}$ is assumed to follow a Poisson distribution:

$$
X_{ij} \sim \operatorname{Poisson}\left((\mathbf{U}\mathbf{V}^T)_{ij}\right)
$$

where:

- $\mathbf{U} \in \mathbb{R}^{n \times k}$ is a non-negative latent factor matrix.
- $\mathbf{V} \in \mathbb{R}^{m \times k}$ is a non-negative latent factor matrix.
- $(\mathbf{U}\mathbf{V}^T)_{ij}$ is the expected rate parameter of the Poisson distribution.

### Characteristics

- **Poisson Noise:** The observed counts are assumed to follow a Poisson distribution.
- **Non-Negativity:** The latent factors $\mathbf{U}$ and $\mathbf{V}$ are non-negative.
- **Optimization:** Parameters can be estimated using maximum likelihood or Bayesian inference.

### Comparison of Factor Analysis Methods

| Factor Analysis Method | Type of Data | Assumptions | Key Characteristics |
|---|---|---|---|
| **PPCA** | Continuous | Gaussian noise | Linear, probabilistic, Gaussian latent variables |
| **Gaussian Matrix Factorization** | Continuous | Gaussian noise | Latent factor model with Gaussian noise |
| **Poisson Matrix Factorization** | Count | Poisson distribution | Non-negative factors, suitable for count data |

### Summary

- **PPCA:** Extends PCA into a probabilistic framework and captures linear relationships with Gaussian noise.
- **Gaussian Matrix Factorization:** Decomposes a data matrix into latent factors assuming Gaussian-distributed errors. It is commonly used in recommendation systems.
- **Poisson Matrix Factorization:** Is designed for count data and decomposes the data matrix into non-negative latent factors using a Poisson distribution.

Each method is suited to different types of data and underlying assumptions.

### Gaussian Matrix Factorization

Gaussian Matrix Factorization is a commonly used technique for collaborative filtering, particularly in recommendation systems. The goal is to factorize a matrix, usually containing user-item ratings, into two lower-dimensional matrices representing latent factors for users and items.

The model assumes that the observed entries are noisy observations from a Gaussian distribution.

### Model Formulation

Given an observed data matrix:

$$
R \in \mathbb{R}^{n \times m}
$$

where:

- $n$ is the number of users.
- $m$ is the number of items.

The goal is to approximate $R$ using two lower-dimensional matrices:

$$
R \approx UV^T
$$

where:

- $U \in \mathbb{R}^{n \times k}$ is the user latent factor matrix.
- $V \in \mathbb{R}^{m \times k}$ is the item latent factor matrix.
- $k$ is the number of latent factors.

The entries of $R$ are assumed to be Gaussian-distributed around $UV^T$.

### Objective Function

The objective is to minimize the squared difference between observed and predicted entries while using regularization to prevent overfitting:

$$
\min_{U,V}
\sum_{(i,j)\in\mathcal{O}}
(R_{ij}-U_iV_j^T)^2
+
\lambda
\left(
\|U\|_F^2+\|V\|_F^2
\right)
$$

where:

- $\mathcal{O}$ is the set of observed entries.
- $\lambda$ is the regularization parameter.
- $\|U\|_F^2$ and $\|V\|_F^2$ are Frobenius norms.

### Solving Using Alternating Least Squares

Alternating Least Squares (ALS) is an iterative optimization technique in which one matrix is fixed while the other is optimized.

### Step 1: Initialize

Initialize $U$ and $V$ with random values or another suitable initialization strategy.

### Step 2: Optimize $U$ Given $V$

For each user $i$, solve:

$$
U_i =
\left(
\sum_{j\in\mathcal{O}_i}
V_jV_j^T
+
\lambda I
\right)^{-1}
\left(
\sum_{j\in\mathcal{O}_i}
R_{ij}V_j
\right)
$$

where $\mathcal{O}_i$ is the set of items rated by user $i$.

### Step 3: Optimize $V$ Given $U$

For each item $j$, solve:

$$
V_j =
\left(
\sum_{i\in\mathcal{O}_j}
U_iU_i^T
+
\lambda I
\right)^{-1}
\left(
\sum_{i\in\mathcal{O}_j}
R_{ij}U_i
\right)
$$

where $\mathcal{O}_j$ is the set of users who rated item $j$.

### Step 4: Iterate

Repeat the updates of $U$ and $V$ until convergence.

A common convergence criterion is that the change in the objective function, $U$, and $V$ becomes smaller than a predefined threshold.

### Gaussian Matrix Factorization in a Probabilistic Context

### Probabilistic Model

In probabilistic Gaussian Matrix Factorization, the observed matrix $R$ is assumed to be generated by a probabilistic process involving latent variables for users and items.

### Model Assumptions

#### 1. Latent Factors

Each user $i$ is associated with a latent vector:

$$
\mathbf{u}_i \in \mathbb{R}^k
$$

Each item $j$ is associated with a latent vector:

$$
\mathbf{v}_j \in \mathbb{R}^k
$$

#### 2. Observed Ratings

The observed rating $R_{ij}$ is modeled as:

$$
R_{ij}
\sim
\mathcal{N}
\left(
\mathbf{u}_i^T\mathbf{v}_j,
\sigma^2
\right)
$$

#### 3. Priors on Latent Factors

The latent factors are assigned Gaussian priors:

$$
\mathbf{u}_i
\sim
\mathcal{N}
\left(
\mathbf{0},
\lambda_u^{-1}I
\right)
$$

and

$$
\mathbf{v}_j
\sim
\mathcal{N}
\left(
\mathbf{0},
\lambda_v^{-1}I
\right)
$$

Here, $\lambda_u$ and $\lambda_v$ are precision parameters, which are the inverse of variance.

### Probabilistic Formulation

The likelihood of the observed data is:

$$
p(R\mid U,V,\sigma^2)
=
\prod_{(i,j)\in\mathcal{O}}
\mathcal{N}
\left(
R_{ij}
\mid
\mathbf{u}_i^T\mathbf{v}_j,
\sigma^2
\right)
$$

The prior distributions are:

$$
p(U\mid\lambda_u)
=
\prod_{i=1}^{n}
\mathcal{N}
\left(
\mathbf{u}_i
\mid
\mathbf{0},
\lambda_u^{-1}I
\right)
$$

and

$$
p(V\mid\lambda_v)
=
\prod_{j=1}^{m}
\mathcal{N}
\left(
\mathbf{v}_j
\mid
\mathbf{0},
\lambda_v^{-1}I
\right)
$$

### Posterior Distribution

The posterior distribution is proportional to the product of the likelihood and the priors:

$$
p(U,V\mid R,\sigma^2,\lambda_u,\lambda_v)
\propto
p(R\mid U,V,\sigma^2)
p(U\mid\lambda_u)
p(V\mid\lambda_v)
$$

### Maximizing the Posterior

Maximizing the posterior is equivalent to minimizing the negative log-posterior:

$$
\mathcal{L}
=
\sum_{(i,j)\in\mathcal{O}}
\frac{
(R_{ij}-\mathbf{u}_i^T\mathbf{v}_j)^2
}{
2\sigma^2
}
+
\frac{\lambda_u}{2}
\sum_{i=1}^{n}
\|\mathbf{u}_i\|^2
+
\frac{\lambda_v}{2}
\sum_{j=1}^{m}
\|\mathbf{v}_j\|^2
$$

The update for $\mathbf{u}_i$ is:

$$
\mathbf{u}_i
=
\left(
\sum_{j\in\mathcal{O}_i}
\mathbf{v}_j\mathbf{v}_j^T
+
\lambda_u\sigma^2I
\right)^{-1}
\left(
\sum_{j\in\mathcal{O}_i}
R_{ij}\mathbf{v}_j
\right)
$$

Similarly, the update for $\mathbf{v}_j$ is:

$$
\mathbf{v}_j
=
\left(
\sum_{i\in\mathcal{O}_j}
\mathbf{u}_i\mathbf{u}_i^T
+
\lambda_v\sigma^2I
\right)^{-1}
\left(
\sum_{i\in\mathcal{O}_j}
R_{ij}\mathbf{u}_i
\right)
$$

### Log-Likelihood

Taking the logarithm of the likelihood:

$$
\log p(R\mid U,V,\sigma^2)
=
-\frac{1}{2\sigma^2}
\sum_{(i,j)\in\mathcal{O}}
(R_{ij}-\mathbf{u}_i^T\mathbf{v}_j)^2
-
\frac{|\mathcal{O}|}{2}
\log(2\pi\sigma^2)
$$

### Priors

The priors on the latent factors are:

$$
p(\mathbf{u}_i)
=
\mathcal{N}
\left(
\mathbf{u}_i
\mid
\mathbf{0},
\lambda_u^{-1}I
\right)
$$

and

$$
p(\mathbf{v}_j)
=
\mathcal{N}
\left(
\mathbf{v}_j
\mid
\mathbf{0},
\lambda_v^{-1}I
\right)
$$

Taking the logarithm:

$$
\log p(\mathbf{u}_i)
=
-\frac{\lambda_u}{2}
\|\mathbf{u}_i\|^2
-
\frac{k}{2}
\log(2\pi\lambda_u^{-1})
$$

and

$$
\log p(\mathbf{v}_j)
=
-\frac{\lambda_v}{2}
\|\mathbf{v}_j\|^2
-
\frac{k}{2}
\log(2\pi\lambda_v^{-1})
$$

### Posterior

The log posterior is:

$$
\log p(U,V\mid R,\sigma^2,\lambda_u,\lambda_v)
=
\log p(R\mid U,V,\sigma^2)
+
\sum_i\log p(\mathbf{u}_i)
+
\sum_j\log p(\mathbf{v}_j)
$$

### Maximum A Posteriori Estimation

The MAP objective is:

$$
\mathcal{L}
=
\frac{1}{2\sigma^2}
\sum_{(i,j)\in\mathcal{O}}
(R_{ij}-\mathbf{u}_i^T\mathbf{v}_j)^2
+
\frac{\lambda_u}{2}
\sum_{i=1}^{n}
\|\mathbf{u}_i\|^2
+
\frac{\lambda_v}{2}
\sum_{j=1}^{m}
\|\mathbf{v}_j\|^2
$$

Taking the partial derivative with respect to $\mathbf{u}_i$:

$$
\frac{\partial\mathcal{L}}
{\partial\mathbf{u}_i}
=
-\frac{1}{\sigma^2}
\sum_{j\in\mathcal{O}_i}
(R_{ij}-\mathbf{u}_i^T\mathbf{v}_j)\mathbf{v}_j
+
\lambda_u\mathbf{u}_i
$$

Setting the derivative to zero:

$$
\sum_{j\in\mathcal{O}_i}
(R_{ij}-\mathbf{u}_i^T\mathbf{v}_j)
\mathbf{v}_j
=
\sigma^2\lambda_u\mathbf{u}_i
$$

Rearranging:

$$
\sum_{j\in\mathcal{O}_i}
R_{ij}\mathbf{v}_j
=
\left(
\sum_{j\in\mathcal{O}_i}
\mathbf{v}_j\mathbf{v}_j^T
+
\sigma^2\lambda_uI
\right)
\mathbf{u}_i
$$

Therefore:

$$
\boxed{
\mathbf{u}_i
=
\left(
\sum_{j\in\mathcal{O}_i}
\mathbf{v}_j\mathbf{v}_j^T
+
\sigma^2\lambda_uI
\right)^{-1}
\left(
\sum_{j\in\mathcal{O}_i}
R_{ij}\mathbf{v}_j
\right)
}
$$

This is the MAP update equation for $\mathbf{u}_i$.

### Non-Negativity in Gaussian Matrix Factorization

Non-negativity is not normally enforced in Gaussian Matrix Factorization. GMF does not inherently require latent factors to be non-negative.

However, non-negativity may be useful when latent factors need to be interpretable or when the application requires positive components.

One common alternative is **Non-Negative Matrix Factorization (NMF)**, which explicitly constrains the latent factors to be non-negative.

Another option is to enforce non-negativity directly during optimization using projected gradient descent.

### Projected Gradient Descent

After a gradient update, project each element onto the non-negative domain.

For $U$:

$$
\mathbf{u}_i
\leftarrow
\mathbf{u}_i
-
\eta
\frac{\partial\mathcal{L}}
{\partial\mathbf{u}_i}
$$

Then:

$$
\mathbf{u}_i
\leftarrow
\max(\mathbf{u}_i,0)
$$

Similarly, for $V$:

$$
\mathbf{v}_j
\leftarrow
\mathbf{v}_j
-
\eta
\frac{\partial\mathcal{L}}
{\partial\mathbf{v}_j}
$$

Then:

$$
\mathbf{v}_j
\leftarrow
\max(\mathbf{v}_j,0)
$$

where $\eta$ is the learning rate.

### Poisson Matrix Factorization

Poisson Matrix Factorization (PMF) is used for decomposing count data into latent factors. The observed entries are assumed to follow a Poisson distribution, and the latent factors are non-negative.

### Probabilistic Model

Given an observed count matrix $R$, we approximate it using:

$$
R \approx UV^T
$$

The model assumes:

$$
R_{ij}
\sim
\operatorname{Poisson}
\left(
(UV^T)_{ij}
\right)
$$

The latent factors satisfy:

$$
\mathbf{u}_i \geq 0
$$

and

$$
\mathbf{v}_j \geq 0
$$

### Objective Function

The Poisson log-likelihood is:

$$
\log p(R\mid U,V)
=
\sum_{(i,j)\in\mathcal{O}}
\left[
R_{ij}
\log((UV^T)_{ij})
-
(UV^T)_{ij}
\right]
-
\sum_{(i,j)\in\mathcal{O}}
\log(R_{ij}!)
$$

### Multiplicative Updates

For fixed $V$, the update for $U$ is:

$$
U_{ik}
\leftarrow
U_{ik}
\frac{
\sum_j
R_{ij}
\frac{V_{jk}}{(UV^T)_{ij}}
}{
\sum_j V_{jk}
}
$$

For fixed $U$, the update for $V$ is:

$$
V_{jk}
\leftarrow
V_{jk}
\frac{
\sum_i
R_{ij}
\frac{U_{ik}}{(UV^T)_{ij}}
}{
\sum_i U_{ik}
}
$$

These multiplicative updates preserve non-negativity provided that $U$ and $V$ are initialized with non-negative values.

### Derivation of the Poisson Matrix Factorization Update

#### Probabilistic Model

The observed counts are modeled as:

$$
R_{ij}
\sim
\operatorname{Poisson}
(\lambda_{ij})
$$

where:

$$
\lambda_{ij}
=
\mathbf{u}_i^T\mathbf{v}_j
$$

#### Log-Likelihood

The likelihood is:

$$
p(R\mid U,V)
=
\prod_{(i,j)\in\mathcal{O}}
\operatorname{Poisson}
(R_{ij}\mid\mathbf{u}_i^T\mathbf{v}_j)
$$

The Poisson probability mass function is:

$$
\operatorname{Poisson}(R_{ij}\mid\lambda_{ij})
=
\frac{
\lambda_{ij}^{R_{ij}}
e^{-\lambda_{ij}}
}{
R_{ij}!
}
$$

Therefore:

$$
\log p(R\mid U,V)
=
\sum_{(i,j)\in\mathcal{O}}
\left[
R_{ij}
\log(\mathbf{u}_i^T\mathbf{v}_j)
-
\mathbf{u}_i^T\mathbf{v}_j
-
\log(R_{ij}!)
\right]
$$

#### Gradient with Respect to $\mathbf{u}_i$

The factorial term is constant with respect to $\mathbf{u}_i$, so it can be ignored during differentiation.

We obtain:

$$
\frac{
\partial\log p(R\mid U,V)
}{
\partial\mathbf{u}_i
}
=
\sum_{j\in\mathcal{O}_i}
\left[
R_{ij}
\frac{\mathbf{v}_j}
{\mathbf{u}_i^T\mathbf{v}_j}
-
\mathbf{v}_j
\right]
$$

or equivalently:

$$
\frac{
\partial\log p(R\mid U,V)
}{
\partial\mathbf{u}_i
}
=
\sum_{j\in\mathcal{O}_i}
\left[
\frac{R_{ij}\mathbf{v}_j}
{\mathbf{u}_i^T\mathbf{v}_j}
-
\mathbf{v}_j
\right]
$$

#### Multiplicative Update

Setting the gradient to zero gives:

$$
\sum_{j\in\mathcal{O}_i}
\frac{R_{ij}\mathbf{v}_j}
{\mathbf{u}_i^T\mathbf{v}_j}
=
\sum_{j\in\mathcal{O}_i}
\mathbf{v}_j
$$

The multiplicative update is:

$$
\boxed{
\mathbf{u}_i
\leftarrow
\mathbf{u}_i
\circ
\frac{
\sum_{j\in\mathcal{O}_i}
R_{ij}
\frac{\mathbf{v}_j}
{\mathbf{u}_i^T\mathbf{v}_j}
}{
\sum_{j\in\mathcal{O}_i}
\mathbf{v}_j
}
}
$$

where $\circ$ denotes element-wise multiplication.

#### Poisson Matrix Factorization with Gamma Priors

Gamma priors are commonly used for non-negative latent factors and can encourage sparsity.

### Probabilistic Model

The observed counts are:

$$
R_{ij}
\sim
\operatorname{Poisson}
(\mathbf{u}_i^T\mathbf{v}_j)
$$

The latent factors have Gamma priors:

$$
u_{ik}
\sim
\operatorname{Gamma}
(\alpha_u,\beta_u)
$$

and:

$$
v_{jk}
\sim
\operatorname{Gamma}
(\alpha_v,\beta_v)
$$

The Gamma density is:

$$
\operatorname{Gamma}(x\mid\alpha,\beta)
=
\frac{
\beta^\alpha
x^{\alpha-1}
e^{-\beta x}
}{
\Gamma(\alpha)
}
$$

where $\alpha$ is the shape parameter and $\beta$ is the rate parameter.

### Joint Log-Likelihood

Ignoring constants independent of $U$ and $V$:

$$
\log p(R,U,V)
=
\sum_{(i,j)\in\mathcal{O}}
\left[
R_{ij}
\log(\mathbf{u}_i^T\mathbf{v}_j)
-
\mathbf{u}_i^T\mathbf{v}_j
\right]
$$

$$
+
\sum_{i,k}
\left[
(\alpha_u-1)\log u_{ik}
-
\beta_u u_{ik}
\right]
$$

$$
+
\sum_{j,k}
\left[
(\alpha_v-1)\log v_{jk}
-
\beta_v v_{jk}
\right]
$$

### Gradient with Respect to $u_{ik}$

The derivative is:

$$
\frac{
\partial\log p(R,U,V)
}{
\partial u_{ik}
}
=
\sum_{j\in\mathcal{O}_i}
\left[
R_{ij}
\frac{v_{jk}}
{\mathbf{u}_i^T\mathbf{v}_j}
-
v_{jk}
\right]
+
\frac{\alpha_u-1}{u_{ik}}
-
\beta_u
$$

Similarly, for $v_{jk}$:

$$
\frac{
\partial\log p(R,U,V)
}{
\partial v_{jk}
}
=
\sum_{i\in\mathcal{O}_j}
\left[
R_{ij}
\frac{u_{ik}}
{\mathbf{u}_i^T\mathbf{v}_j}
-
u_{ik}
\right]
+
\frac{\alpha_v-1}{v_{jk}}
-
\beta_v
$$

### Multiplicative Updates with Gamma Priors

A commonly used multiplicative update form for $U$ is:

$$
u_{ik}
\leftarrow
u_{ik}
\frac{
\displaystyle
\sum_{j\in\mathcal{O}_i}
R_{ij}
\frac{v_{jk}}
{\mathbf{u}_i^T\mathbf{v}_j}
+
\alpha_u-1
}{
\displaystyle
\sum_{j\in\mathcal{O}_i}
v_{jk}
+
\beta_u
}
$$

Similarly, for $V$:

$$
v_{jk}
\leftarrow
v_{jk}
\frac{
\displaystyle
\sum_{i\in\mathcal{O}_j}
R_{ij}
\frac{u_{ik}}
{\mathbf{u}_i^T\mathbf{v}_j}
+
\alpha_v-1
}{
\displaystyle
\sum_{i\in\mathcal{O}_j}
u_{ik}
+
\beta_v
}
$$

These updates preserve non-negativity when initialized with positive values.

> **Note:** The exact multiplicative-update form depends on the optimization derivation and parameterization of the Gamma prior. The equations above correspond to a common gradient-based construction. A fully Bayesian treatment instead uses variational inference or Gibbs sampling.

### Poisson Matrix Factorization with Gamma Priors and ELBO

A more principled Bayesian approach to Poisson Matrix Factorization with Gamma priors uses variational inference.

### Variational Inference with Auxiliary Variables

Variational inference approximates the true posterior with a simpler distribution that can be optimized efficiently.

The model is:

$$
R_{ij}
\sim
\operatorname{Poisson}
(\mathbf{u}_i^T\mathbf{v}_j)
$$

with priors:

$$
u_{ik}
\sim
\operatorname{Gamma}
(\alpha_u,\beta_u)
$$

and:

$$
v_{jk}
\sim
\operatorname{Gamma}
(\alpha_v,\beta_v)
$$

### Variational Distribution

We use a mean-field approximation:

$$
q(U,V)
=
\prod_{i,k}
q(u_{ik})
\prod_{j,k}
q(v_{jk})
$$

where:

$$
q(u_{ik})
=
\operatorname{Gamma}
(
\tilde{\alpha}_{u_{ik}},
\tilde{\beta}_{u_{ik}}
)
$$

and:

$$
q(v_{jk})
=
\operatorname{Gamma}
(
\tilde{\alpha}_{v_{jk}},
\tilde{\beta}_{v_{jk}}
)
$$

The variational parameters are:

- $\tilde{\alpha}$: variational shape parameter.
- $\tilde{\beta}$: variational rate parameter.

### Evidence Lower Bound

The Evidence Lower Bound (ELBO) is:

$$
\operatorname{ELBO}
=
\mathbb{E}_q
[
\log p(R,U,V)
]
-
\mathbb{E}_q
[
\log q(U,V)
]
$$

Maximizing the ELBO makes the variational approximation closer to the true posterior.

### Expected Log-Likelihood

The expected log-likelihood is:

$$
\mathbb{E}_q
[
\log p(R\mid U,V)
]
=
\sum_{(i,j)\in\mathcal{O}}
\mathbb{E}_q
\left[
R_{ij}
\log(\mathbf{u}_i^T\mathbf{v}_j)
-
\mathbf{u}_i^T\mathbf{v}_j
\right]
+
\text{const}
$$

The expectation of the rate parameter is:

$$
\mathbb{E}_q
[
\mathbf{u}_i^T\mathbf{v}_j
]
=
\sum_k
\mathbb{E}_q[u_{ik}]
\mathbb{E}_q[v_{jk}]
$$

For a Gamma random variable:

$$
\mathbb{E}[u_{ik}]
=
\frac{
\tilde{\alpha}_{u_{ik}}
}{
\tilde{\beta}_{u_{ik}}
}
$$

and:

$$
\mathbb{E}[v_{jk}]
=
\frac{
\tilde{\alpha}_{v_{jk}}
}{
\tilde{\beta}_{v_{jk}}
}
$$

Therefore:

$$
\mathbb{E}_q
[
\mathbf{u}_i^T\mathbf{v}_j
]
=
\sum_k
\frac{
\tilde{\alpha}_{u_{ik}}
}{
\tilde{\beta}_{u_{ik}}
}
\frac{
\tilde{\alpha}_{v_{jk}}
}{
\tilde{\beta}_{v_{jk}}
}
$$

### Prior Term

For $U$:

$$
\mathbb{E}_q[\log p(U)]
=
\sum_{i,k}
\left[
(\alpha_u-1)
\mathbb{E}_q[\log u_{ik}]
-
\beta_u
\mathbb{E}_q[u_{ik}]
\right]
+
\text{const}
$$

For a Gamma distribution:

$$
\mathbb{E}_q[\log u_{ik}]
=
\psi(
\tilde{\alpha}_{u_{ik}}
)
-
\log
\tilde{\beta}_{u_{ik}}
$$

and:

$$
\mathbb{E}_q[u_{ik}]
=
\frac{
\tilde{\alpha}_{u_{ik}}
}{
\tilde{\beta}_{u_{ik}}
}
$$

Therefore:

$$
\mathbb{E}_q[\log p(U)]
=
\sum_{i,k}
\left[
(\alpha_u-1)
\left(
\psi(\tilde{\alpha}_{u_{ik}})
-
\log\tilde{\beta}_{u_{ik}}
\right)
-
\beta_u
\frac{
\tilde{\alpha}_{u_{ik}}
}{
\tilde{\beta}_{u_{ik}}
}
\right]
+
\text{const}
$$

A corresponding expression applies to $V$.

### Entropy Term

The entropy contribution can be written through:

$$
-\mathbb{E}_q[\log q(U)]
$$

and:

$$
-\mathbb{E}_q[\log q(V)]
$$

For the Gamma variational distribution:

$$
\mathbb{E}_q[\log u_{ik}]
=
\psi(
\tilde{\alpha}_{u_{ik}}
)
-
\log
\tilde{\beta}_{u_{ik}}
$$

and:

$$
\mathbb{E}_q[u_{ik}]
=
\frac{
\tilde{\alpha}_{u_{ik}}
}{
\tilde{\beta}_{u_{ik}}
}
$$

These quantities are used to evaluate the ELBO.

### Variational Updates

The variational parameters are updated iteratively.

A useful quantity for the expected contribution of component $k$ to observation $(i,j)$ is:

$$
\phi_{ijk}
=
\frac{
\mathbb{E}_q[u_{ik}]
\mathbb{E}_q[v_{jk}]
}{
\sum_{\ell}
\mathbb{E}_q[u_{i\ell}]
\mathbb{E}_q[v_{j\ell}]
}
$$

where:

$$
\sum_k \phi_{ijk}=1
$$

The expected number of observations assigned to component $k$ is:

$$
\hat{R}_{ijk}
=
R_{ij}\phi_{ijk}
$$

### Update for $u_{ik}$

The variational shape parameter becomes:

$$
\tilde{\alpha}_{u_{ik}}
=
\alpha_u
+
\sum_{j\in\mathcal{O}_i}
\hat{R}_{ijk}
$$

The variational rate parameter becomes:

$$
\tilde{\beta}_{u_{ik}}
=
\beta_u
+
\sum_{j\in\mathcal{O}_i}
\mathbb{E}_q[v_{jk}]
$$

Therefore:

$$
q(u_{ik})
=
\operatorname{Gamma}
(
\tilde{\alpha}_{u_{ik}},
\tilde{\beta}_{u_{ik}}
)
$$

with posterior mean:

$$
\mathbb{E}_q[u_{ik}]
=
\frac{
\tilde{\alpha}_{u_{ik}}
}{
\tilde{\beta}_{u_{ik}}
}
$$

### Update for $v_{jk}$

Similarly:

$$
\tilde{\alpha}_{v_{jk}}
=
\alpha_v
+
\sum_{i\in\mathcal{O}_j}
\hat{R}_{ijk}
$$

and:

$$
\tilde{\beta}_{v_{jk}}
=
\beta_v
+
\sum_{i\in\mathcal{O}_j}
\mathbb{E}_q[u_{ik}]
$$

Therefore:

$$
q(v_{jk})
=
\operatorname{Gamma}
(
\tilde{\alpha}_{v_{jk}},
\tilde{\beta}_{v_{jk}}
)
$$

with posterior mean:

$$
\mathbb{E}_q[v_{jk}]
=
\frac{
\tilde{\alpha}_{v_{jk}}
}{
\tilde{\beta}_{v_{jk}}
}
$$

### Solving for the Variational Parameters

For $u_{ik}$, the variational distribution is:

$$
q(u_{ik})
=
\operatorname{Gamma}
(
\tilde{\alpha}_{u_{ik}},
\tilde{\beta}_{u_{ik}}
)
$$

The expected value is:

$$
\mathbb{E}_q[u_{ik}]
=
\frac{
\tilde{\alpha}_{u_{ik}}
}{
\tilde{\beta}_{u_{ik}}
}
$$

and the expected logarithm is:

$$
\mathbb{E}_q[\log u_{ik}]
=
\psi(
\tilde{\alpha}_{u_{ik}}
)
-
\log
\tilde{\beta}_{u_{ik}}
$$

The corresponding expressions for $v_{jk}$ are:

$$
\mathbb{E}_q[v_{jk}]
=
\frac{
\tilde{\alpha}_{v_{jk}}
}{
\tilde{\beta}_{v_{jk}}
}
$$

and:

$$
\mathbb{E}_q[\log v_{jk}]
=
\psi(
\tilde{\alpha}_{v_{jk}}
)
-
\log
\tilde{\beta}_{v_{jk}}
$$

### Expected Rate

The expected Poisson rate is:

$$
\mathbb{E}_q[\lambda_{ij}]
=
\sum_k
\mathbb{E}_q[u_{ik}]
\mathbb{E}_q[v_{jk}]
$$

or:

$$
\mathbb{E}_q[\lambda_{ij}]
=
\sum_k
\frac{
\tilde{\alpha}_{u_{ik}}
}{
\tilde{\beta}_{u_{ik}}
}
\frac{
\tilde{\alpha}_{v_{jk}}
}{
\tilde{\beta}_{v_{jk}}
}
$$

This expected rate is used when updating the auxiliary responsibility variables $\phi_{ijk}$.

### Complete Variational Inference Algorithm

The overall algorithm can be summarized as follows.

### Step 1: Initialize

Initialize positive variational parameters:

$$
\tilde{\alpha}_{u_{ik}},
\tilde{\beta}_{u_{ik}},
\tilde{\alpha}_{v_{jk}},
\tilde{\beta}_{v_{jk}}
$$

### Step 2: Compute Expected Latent Factors

Compute:

$$
\mathbb{E}[u_{ik}]
=
\frac{
\tilde{\alpha}_{u_{ik}}
}{
\tilde{\beta}_{u_{ik}}
}
$$

and:

$$
\mathbb{E}[v_{jk}]
=
\frac{
\tilde{\alpha}_{v_{jk}}
}{
\tilde{\beta}_{v_{jk}}
}
$$

### Step 3: Compute Responsibilities

For each observed entry:

$$
\phi_{ijk}
=
\frac{
\mathbb{E}[u_{ik}]
\mathbb{E}[v_{jk}]
}{
\sum_{\ell}
\mathbb{E}[u_{i\ell}]
\mathbb{E}[v_{j\ell}]
}
$$

### Step 4: Compute Expected Counts

$$
\hat{R}_{ijk}
=
R_{ij}\phi_{ijk}
$$

### Step 5: Update $U$

Update:

$$
\tilde{\alpha}_{u_{ik}}
=
\alpha_u
+
\sum_{j\in\mathcal{O}_i}
\hat{R}_{ijk}
$$

and:

$$
\tilde{\beta}_{u_{ik}}
=
\beta_u
+
\sum_{j\in\mathcal{O}_i}
\mathbb{E}[v_{jk}]
$$

### Step 6: Update $V$

Update:

$$
\tilde{\alpha}_{v_{jk}}
=
\alpha_v
+
\sum_{i\in\mathcal{O}_j}
\hat{R}_{ijk}
$$

and:

$$
\tilde{\beta}_{v_{jk}}
=
\beta_v
+
\sum_{i\in\mathcal{O}_j}
\mathbb{E}[u_{ik}]
$$

### Step 7: Repeat

Repeat Steps 2–6 until the ELBO or another convergence criterion stabilizes.

### Summary

The major factorization methods discussed in this chapter differ primarily in their assumptions about the observed data.

- **PPCA** uses a probabilistic model with Gaussian noise and linear latent variables.
- **Gaussian Matrix Factorization** models continuous observations using Gaussian noise and latent user/item factors.
- **Poisson Matrix Factorization** models count observations using Poisson distributions and non-negative latent factors.
- **Gamma Priors** provide a natural Bayesian prior for non-negative latent factors.
- **Variational Inference** approximates the posterior using a tractable family of distributions.
- **The ELBO** provides the objective that is optimized during variational inference.
- **Auxiliary variables** allow the Poisson likelihood to be handled using simpler conditional updates.
- **Gamma variational distributions** provide closed-form updates for the variational shape and rate parameters.

The key distinction is the observation model:

$$
\boxed{
\text{Continuous data}
\rightarrow
\text{Gaussian models}
}
$$

and:

$$
\boxed{
\text{Count data}
\rightarrow
\text{Poisson models}
}
$$

For Bayesian Poisson Matrix Factorization:

$$
\boxed{
\text{Poisson likelihood}
+
\text{Gamma priors}
+
\text{Variational inference}
\rightarrow
\text{Gamma posterior approximations}
}
$$

This framework provides a flexible approach for modeling sparse count data while maintaining non-negative and interpretable latent representations.
