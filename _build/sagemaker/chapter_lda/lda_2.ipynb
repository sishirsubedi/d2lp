# LDA 2
:label:`sec_lda2`

Probabilistic Latent Semantic Analysis (PLSA) and Latent Dirichlet Allocation (LDA) are both probabilistic models used for topic modeling, which is a method to discover abstract topics in a collection of documents. LDA is considered an improvement over PLSA in several significant ways.

### Generative Process

PLSA: PLSA models each document as a mixture of topics, but it does not have a generative process for documents. The mixture weights are learned directly from the training data, meaning the model does not naturally generalize to new, unseen documents. Specifically, PLSA does not define a probability distribution over the topic mixtures for the entire collection, making it challenging to predict for new documents.

LDA: LDA introduces a generative process, where each document is assumed to be generated by first choosing a distribution over topics, drawn from a Dirichlet distribution. Then each word in the document is generated by first choosing a topic according to the document's topic distribution and then choosing a word according to the topic's word distribution. This process allows LDA to define a complete generative model, including priors, making it possible to generalize to new documents.

### Handling New Documents

PLSA: The model parameters include a topic distribution for each document in the training set, but it does not provide a mechanism for assigning topics to new, unseen documents because it lacks a prior distribution over topics.

LDA: LDA addresses this limitation by using Dirichlet priors for the distribution of topics over documents and words over topics. This allows LDA to infer the topic distribution for new documents using Bayesian inference.

### Parameter Estimation and Regularization

PLSA: PLSA is prone to overfitting because it lacks a prior distribution, and the parameters are estimated directly from the data without regularization.

LDA: LDA incorporates Dirichlet priors, which act as regularizers. The Dirichlet prior helps prevent overfitting by smoothing the distribution of topics over documents and words over topics. This regularization makes LDA more robust and capable of providing more coherent topics.

### Interpretability and Coherence of Topics

PLSA: Topics generated by PLSA may not always be as coherent or interpretable because the model is more prone to overfitting and does not include a mechanism for encouraging sparsity in the topic distributions.

LDA: The use of Dirichlet priors in LDA encourages sparsity in the topic distributions, leading to more interpretable and coherent topics. This means that each document is typically associated with a small number of topics, and each topic is characterized by a small number of words with high probabilities.

### LDA Parameter Estimation

In Latent Dirichlet Allocation (LDA), the parameters $p(w\vert{}z)$, the probability of word $w$ given topic $z$, and $p(z\vert{}d)$, the probability of topic $z$ given document $d$, are estimated through the generative model and an inference algorithm such as Variational Inference or Gibbs Sampling.

### Generative Model

LDA's generative process is:

For each topic $z$, draw a distribution over words $\phi_z$ from a Dirichlet distribution with hyperparameter $\beta$:

$$\phi_z \sim \text{Dir}(\beta)$$

For each document $d$, draw a distribution over topics $\theta_d$ from a Dirichlet distribution with hyperparameter $\alpha$:

$$\theta_d \sim \text{Dir}(\alpha)$$

For each word $w$ in document $d$:

Draw a topic $z_{d,n}$ from the multinomial distribution $\theta_d$.

Draw a word $w_{d,n}$ from the multinomial distribution $\phi_{z_{d,n}}$.

The parameters $p(w\vert{}z)$ and $p(z\vert{}d)$ correspond to $\phi_z$ and $\theta_d$.

1. Estimating $p(w\vert{}z)$

For each topic $z$, $\phi_z$ represents the distribution over words.

Using Gibbs Sampling, an estimate of $\phi_z$ can be computed as:

$$\phi_{w\vert{}z} = \frac{n_{w,z} + \beta}{\sum_{w'} (n_{w',z} + \beta)}$$

where $n_{w,z}$ is the count of word $w$ assigned to topic $z$ across the corpus, and $\beta$ is the Dirichlet prior.

Interpretation:

$\phi_{w\vert{}z}$ gives the probability of observing word $w$ given topic $z$.

3. Estimating $p(z\vert{}d)$

For each document $d$, $\theta_d$ represents the distribution over topics.

Using Gibbs Sampling, an estimate of $\theta_d$ can be computed as:

$$\theta_{z\vert{}d} = \frac{n_{z,d} + \alpha}{\sum_{z'} (n_{z',d} + \alpha)}$$

where $n_{z,d}$ is the count of words in document $d$ assigned to topic $z$, and $\alpha$ is the Dirichlet prior.

Interpretation:

$\theta_{z\vert{}d}$ gives the probability of topic $z$ being present in document $d$.

Summary

In LDA, unlike PLSA, the parameters $p(w\vert{}z)$ and $p(z\vert{}d)$ are not directly calculated from data counts alone. They are inferred as part of a probabilistic model that includes Dirichlet priors.

Inference can be performed using Variational Inference or Gibbs Sampling. The Dirichlet priors $\alpha$ and $\beta$ help regularize the estimates and allow the model to generalize to new documents.

The resulting parameters $\phi_z$ and $\theta_d$ provide the probabilities of words given topics and topics given documents.

### LDA vs PLSA: Formulaic Comparison

LDA is an extension and improvement over Probabilistic Latent Semantic Analysis (PLSA). The main difference is that LDA introduces Dirichlet priors over the topic distributions.

1. Model Definition and Generative Process

PLSA Model

In PLSA, the probability of a word given a topic is:

$$p(w\vert{}z) = \phi_{w\vert{}z}$$

where $\phi_{w\vert{}z}$ represents the probability of word $w$ given topic $z$.

The probability of a topic given a document is:

$$p(z\vert{}d) = \theta_{z\vert{}d}$$

where $\theta_{z\vert{}d}$ represents the probability of topic $z$ given document $d$.

The joint probability of a word and a document is:

$$p(d,w) = p(d) \sum_z p(z\vert{}d) p(w\vert{}z)$$

PLSA does not define a prior distribution over $\theta_{z\vert{}d}$. Therefore, $\theta_{z\vert{}d}$ is only directly defined for documents in the training set.

LDA Model

In LDA, prior distributions are introduced over the document-topic and topic-word distributions.

Document-topic distribution:

$$\theta_d \sim \text{Dir}(\alpha)$$

Topic-word distribution:

$$\phi_z \sim \text{Dir}(\beta)$$

The probability of a word given a topic is:

$$p(w\vert{}z) = \phi_{w\vert{}z}$$

The probability of a topic given a document is:

$$p(z\vert{}d) = \theta_{z\vert{}d}$$

The joint probability can be written as:

$$p(d,w,z,\theta_d,\phi_z) = p(d) p(\theta_d\vert{}\alpha) p(z\vert{}\theta_d) p(w\vert{}z,\phi_z)$$

### Formulaic Differences and Implications

Parameter Estimation and Regularization

PLSA directly estimates $\phi_{w\vert{}z}$ and $\theta_{z\vert{}d}$ from data without priors, which can lead to overfitting.

LDA uses Dirichlet priors $\alpha$ and $\beta$ over $\theta_d$ and $\phi_z$.

The Dirichlet prior for $\theta_d$ is:

$$p(\theta_d\vert{}\alpha) = \frac{\Gamma(\sum_i \alpha_i)}{\prod_i \Gamma(\alpha_i)} \prod_i \theta_{d,i}^{\alpha_i - 1}$$

The Dirichlet prior for $\phi_z$ is:

$$p(\phi_z\vert{}\beta) = \frac{\Gamma(\sum_i \beta_i)}{\prod_i \Gamma(\beta_i)} \prod_i \phi_{z,i}^{\beta_i - 1}$$

These priors act as regularizers, helping prevent overfitting and allowing the model to generalize better to new data.

Inference and Generalization

PLSA cannot easily handle new documents because it does not define a generative process for document-topic distributions.

LDA can infer the topic distribution $\theta_d$ for a new document using the posterior:

$$p(\theta_d\vert{}w_d,\alpha,\beta) \propto p(\theta_d\vert{}\alpha) \prod_{n=1}^{N_d} \sum_{z=1}^K p(z\vert{}\theta_d) p(w_n\vert{}z,\beta)$$

This is possible because LDA includes the Dirichlet prior $\alpha$ on the document-topic distribution $\theta_d$.

Summary

LDA's inclusion of Dirichlet priors $\alpha$ and $\beta$ provides a more principled framework than PLSA.

The main advantages of LDA over PLSA are:

LDA has a complete generative probabilistic model.

LDA uses Dirichlet priors for regularization.

LDA is better able to generalize to new documents.

LDA can produce sparse and interpretable topic distributions.

LDA is less prone to overfitting than PLSA.

###  Model Complexity

PLSA: PLSA has a simpler model structure without priors on the topic distributions. The EM algorithm used in PLSA calculates posterior probabilities $p(z\vert{}d,w)$ and updates the model parameters $p(w\vert{}z)$ and $p(z\vert{}d)$. The absence of priors means fewer parameters and simpler updates.

LDA: LDA introduces Dirichlet priors for both the document-topic distribution $\theta_d$ and the topic-word distribution $\phi_z$. The inference process, such as Variational Inference or Gibbs Sampling, requires iterative updates of the variational parameters or topic assignments. The additional complexity can make LDA computationally more intensive.

- Inference Algorithm

PLSA: The EM algorithm used in PLSA generally has straightforward updates that converge relatively quickly. However, PLSA can suffer from overfitting and lacks a probabilistic generative process for new documents.

LDA: Variational Inference in LDA involves optimizing variational parameters, which can be computationally expensive because of the need to calculate digamma functions and ensure convergence of the variational lower bound. Gibbs Sampling involves iteratively sampling topic assignments for each word, which can also be slow for large datasets.

- Convergence and Scalability

PLSA: Convergence criteria in PLSA are typically based on changes in the likelihood function. PLSA may converge faster because fewer parameters are updated.

LDA: Convergence criteria in LDA involve the variational lower bound or changes in the latent-variable distributions. The presence of priors and the need to maintain a coherent generative model can result in slower convergence, particularly for large datasets.

### Final Summary

While LDA generally provides a more robust and principled approach to topic modeling, especially for handling new documents and providing interpretable topics, it can be computationally slower than PLSA because of the complexity of the model and the inference algorithms used.

The specific implementation and optimization techniques can significantly affect the relative speed of both methods. In practice, efficient implementations or approximations can be used to balance accuracy and computational efficiency.
